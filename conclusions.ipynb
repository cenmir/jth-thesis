## Answered research questions

**RQ1: How do the rise time, settling time, overshoot and steady-state error of the bio-inspired adaptive PID controller compare with those of a Ziegler-Nichols-tuned fixed-gain controller, across an operating envelope spanning 0 to 1 m/s forward velocity?**

The adaptive controller improved every measured metric. Rise time fell by 26 %, settling time by 40 %, overshoot by 52 %, and steady-state error by a factor of 3.75 (see @sec-analysis). Run-to-run variability of settling time more than halved.

**RQ2: By how much does the adaptive controller reduce the peak tilt angle and the recovery time after a 15° impulse disturbance, compared with the fixed-gain controller, when both run at 200 Hz on the target MCU?**

The peak tilt angle after the 15° impulse fell from 9.8° (fixed-gain) to 6.2° (adaptive), a 37 % reduction. Mean recovery time to within $|\theta| < 5°$ fell from 1.71 s to 0.84 s, a 51 % reduction. All ten adaptive runs returned to threshold within 1.1 s, while three of ten fixed-gain runs exceeded the recovery threshold at the end of the 2 s observation window.
