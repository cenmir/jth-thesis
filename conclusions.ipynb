## Answered research questions

**RQ1: Does a real-time adaptive PID controller tuned with a bio-inspired optimization algorithm provide significantly better control performance than a fixed-gain PID controller?**

Yes. The adaptive variant improved every measured metric, with a 40 % reduction in settling time and a 50 % reduction in overshoot, as detailed in @sec-analysis.

**RQ2: Does adaptive PID tuning improve the robot's ability to reject external disturbances and handle parameter variations compared to a conventional PID controller?**

Yes. The adaptive controller stayed within the recovery threshold on all ten impulse-disturbance trials, while the fixed-gain controller exceeded the threshold on three of ten.
