# Introduction {#sec-introduction}

*This chapter introduces the research area, frames the problem, states the purpose and research questions, defines the delimitations, and ends with the outline of the report. It tells the reader why the work matters and what the rest of the thesis sets out to do.*


## Background

*Provide enough domain context for a reader outside the speciality to understand the problem. Cite the foundational and most recent work that situates the study. Do not write a textbook; cover the field broadly enough that the reader can place the work and trust that the relevant literature has been read.*

Example, drawn from a thesis on a two-wheeled self-balancing robot. The first paragraph anchors the field; the second narrows toward the specific class of robot; the third points at the controller problem the thesis will tackle.

> The development of robotics has come a long way, especially for mobile robots [@Braunl2022]. Robots that can follow human commands, move autonomously, or run preprogrammed sequences are called mobile robots [@Wang2025]. Previously, mobile robots were operated by heavy, large and expensive computer systems that were not portable. Modern mobile robots are built with numerous motors and sensors controlled by inexpensive, small and light embedded computers on board.
>
> Within mobile robotics the two-wheeled robot (TWR) is one of the least popular designs because of its structural imbalance and control complexity [@Mathew2021]. To enable self-balancing, TWRs are equipped with inertial measurement units (IMUs) and run advanced control algorithms, typically feedback control, to maintain stability [@Arefin2023].
>
> The classical Proportional-Integral-Derivative (PID) controller has been used on numerous robots, but it is tuned for a single operating point and degrades when external disturbances or parameter variations occur. Adaptive control structures modify their parameters in real time to maintain the desired performance [@Olmez2022; @Paulescu2021], and several have been proposed for stabilising inverted pendulums.

Notice three things in the example. First, every claim that is not common knowledge carries a citation. Second, multiple citations to the same claim use the `[@a; @b]` form. Third, sentences are short and concrete, with no filler phrases.


## Problem description

*State the concrete problem the thesis addresses. The reader should be able to point at the gap, the bottleneck or the unmet need after reading this section. A common pattern: a paragraph naming the problem, a paragraph on why it matters now, and a paragraph on why existing solutions fall short.*

Keep the problem statement narrow enough that the rest of the thesis can plausibly answer it. A useful test: if a reader cannot, from the problem description alone, predict the shape of the research questions, the problem is too broad or too vague.


## Requirements

*If the thesis is structured around requirements (typical for engineering projects with an industry partner), list them here. Each requirement should be measurable, with units, so that the results chapter can decide unambiguously whether the requirement is met.*

Example of well-formed requirements:

1. The controller shall keep the pitch angle within $|\theta| < 5°$ during steady-state operation on a level surface.
2. The robot shall recover to within $|\theta| < 5°$ within 2 s after an impulse disturbance of $\pm 15°$.
3. The control loop shall run at no less than 200 Hz on the target MCU.
4. Total bill-of-materials cost shall not exceed 1 500 SEK.

Avoid qualitative wording ("the robot shall be stable", "the system shall be cost-effective"). Translate stakeholder needs into engineering targets with units.


## Purpose and research questions

*State the purpose in one or two sentences. Then formulate research questions (RQs) that, taken together, must be answered to fulfil the purpose. Number the RQs. Every RQ must be answered explicitly in the Conclusions chapter.*

Example:

> The purpose of this thesis is to quantify the control performance and disturbance rejection of a real-time adaptive PID controller, tuned by a bio-inspired optimization algorithm, on a two-wheeled self-balancing robot, in comparison with a conventional fixed-gain PID controller.
>
> **RQ1:** How do the rise time, settling time, overshoot and steady-state error of the bio-inspired adaptive PID controller compare with those of a Ziegler-Nichols-tuned fixed-gain controller, across an operating envelope spanning 0 to 1 m/s forward velocity?
>
> **RQ2:** By how much does the adaptive controller reduce the peak tilt angle and the recovery time after a 15° impulse disturbance, compared with the fixed-gain controller, when both run at 200 Hz on the target MCU?

### Rules for good research questions

Read this section as a manifesto, not a tip box. The bar laid out here applies to both bachelor and master theses written in the School of Engineering at Jönköping University. The bachelor scope is smaller; the rules are the same.

The Intended Learning Outcomes for this course are listed in your course syllabus. The bachelor and master courses use the same themes with slightly different phrasing and numbering, so this template refers to the themes by name (analysis and evaluation of technical solutions, modelling and simulation, sustainable design, and so on) rather than by ILO number. Map each theme to the numbered ILO in your specific syllabus before you reference it in the body.

#### The bar

This is a thesis in the School of Engineering. The programmes are *Industrial Product Realisation* and *Mechanical Engineering, Design*. The bar is engineering work: simulation, mathematics, computation, physical testing, design-for-manufacture (DfM), design-for-assembly (DfA), tolerance analysis, or some combination thereof. Not just CAD. Not just literature. Not just process description. Not just workshops.

Aim for three research questions. Two is sometimes too thin; four is often a sign that the scope was never sharpened. An RQ that does not demand engineering work is an RQ that does not belong here.

A good RQ passes all four tests:

1. **Specific, not general.** "How can the product be improved?" is vague. Specify what aspect, by what measure, under what conditions.
2. **Not answerable by yes or no alone.** "Does X outperform Y?" or "Can X be simulated?" collapses to yes/no with no engineering depth. Reformulate as "How does X compare to Y, measured by ..." or "How can X be simulated, and what accuracy is achieved against the experimental benchmark?".
3. **Requires engineering development work.** Simulation, physical testing, prototyping, optimisation, mathematical analysis or DfM/DfA. A question answerable by literature review or by stakeholder workshops alone is insufficient.
4. **Answerable within scope.** If the question needs resources or time beyond the thesis project, narrow it.

A common failure mode is the *overpromised RQ on solid work*: the question asks "what factors cause X" when the answer turns out to be intuitive or already in the textbook. If this happens, reformulate the RQ rather than penalise the work. Either demote the soft work to method (user study, need-finding) and free up the RQ slot for a question with engineering depth, or rewrite the RQ to demand the quantitative analysis the work actually does ("how do X and Y affect Z, and at what threshold does Z fail?").

#### What counts as engineering work

The thesis must demonstrate at least one of the following in real depth. Most theses combine two or three. A thesis that combines none is not a thesis from this school.

- **Simulation and modeling.** FEA, CFD, multi-body dynamics, system simulation, control-loop simulation. Must be validated against measurements or against an analytical benchmark.
- **Mathematical derivation and analysis.** Free-body diagrams, governing equations, parameter studies, sensitivity analysis, dimensional analysis, analytical estimates.
- **Computation.** Algorithms implemented and verified against test cases, scripted parameter sweeps, numerical optimisation, data processing pipelines.
- **Physical testing.** Real measurements on real hardware, with stated equipment make and model, calibration, sample size, statistical treatment, and uncertainty analysis.
- **Manufacturing- and assembly-oriented design.** DfM, DfA, tolerance analysis, manufacturability assessment tied to a specific process (turning, milling, casting, deep drawing, injection moulding, additive manufacturing).
- **Systematic concept selection.** Pugh matrix or weighted decision matrix with explicit, measurable criteria and weighted scores. Not "we picked the one the company liked".

Every thesis must validate. A product-development thesis that proposes three concepts and stops there is unfinished. A simulation thesis without comparison to measurement or analytical limit is unfinished. A testing thesis with raw data and no analysis is unfinished.

#### What this thesis must not be

A few patterns recur often enough to warrant naming:

- **The paper product.** Concepts on slides, CAD renderings, glossy mock-ups, no analysis, no measurement, no test. This is a maker hobby with academic formatting. If a high-schooler with SolidWorks could produce the same artefact, the thesis does not fulfil the ILOs on technical-solution analysis and on modelling, simulation or prediction. Build it, simulate it, test it, or cut it.
- **The CAD-only thesis.** A 3D model, some renderings, the words "this could be 3D-printed". No FEA, no calculations, no manufacturability analysis, no test. CAD is the *output* of engineering thinking, not the substitute for it.
- **The tons-of-text thesis.** A bloated literature review and a chapter on research philosophy (deductive, inductive, qualitative, quantitative, credibility, transferability, dependability, confirmability) bolted onto thin engineering work. Word count is not engineering depth. The reader should learn more about the design and *why it works* than about the methodology of writing about design.
- **The "the company asked us to" thesis.** A sequence of design choices justified by stakeholder preference, not engineering analysis. Stakeholders can be wrong; show why the choice is right, with numbers.
- **The 200-page workshop log.** Interview transcripts, brainstorming output, journey maps. Useful as input. Insufficient as output. The thesis is what was built or analysed *because of* the workshops, not the workshops themselves.

#### Pointing back to the ILOs

The Intended Learning Outcomes are the formal criteria the examiner applies. The themes most often at risk are these. Find the matching numbered ILO in your own course syllabus before you cite one in the body of the thesis.

- **Analysis and evaluation of technical solutions.** Something is *analysed and evaluated* with depth and well-supported arguments. Not just proposed.
- **Planning and execution with adequate methods.** A clear plan, executed with methods appropriate to the question, within the stated time and resource frame.
- **Modelling, simulation, prediction, evaluation.** Claims about how a system behaves are supported by simulation, measurement, or analytical estimate. Not by intuition.
- **Design with sustainability considerations.** Products, processes, or systems are designed with regard to economic, social and ecological sustainability, in concrete terms (mass saved, energy avoided, lifetime extended), not as a list of UN goals.
- **Written and oral communication.** Reproducible, precise, engineering in voice. The reader can rebuild the work from the report.
- **Scientific, societal and ethical judgement.** Reflection on the wider implications of the work, with specific reasoning, not platitudes.

If, after reading the thesis, an external engineer cannot point at the moment where engineering analysis happened, the ILOs on technical-solution analysis and on modelling, simulation or prediction are not met. No amount of process description, literature review, or stakeholder workshop can substitute for that moment. Find it, do it, write it up.

#### Bachelor versus master scope

The scope differs in depth and breadth, not in kind. A bachelor thesis can be a single, sharp investigation: one component, one load case, one measurement campaign. A master thesis is expected to chain two or three of those into a coherent analysis or to push the depth (more material models, more parameter studies, more validation). In both, the rules above apply. A bachelor thesis with a paper product fails the same way a master thesis does.


## Delimitations

*State what the thesis does **not** cover. Delimitations are deliberate scope choices made by the authors, not external constraints. Each delimitation should reduce reader confusion about what to expect from the rest of the report.*

Examples that work well:

- The study is limited by the hardware used in the robot. The computing capacity of the control board, the accuracy and sampling rate of the sensors, and the response speed and torque of the motors are taken as fixed and not optimised further.
- The work focuses only on control and stabilisation. Other aspects of mobile robotics, such as navigation, path planning and obstacle avoidance, are not considered.
- All experiments are performed indoors on a smooth, level surface. Uneven terrain and outdoor environments are not evaluated.

Distinguish delimitations (choices) from limitations (constraints discovered along the way). Limitations belong in the discussion chapter.

### Do not delimit away the engineering work

A delimitation is a narrowing of scope, not a way to dodge the engineering bar laid out in section 1.4.1. "We delimit ourselves from performing simulations" is not a delimitation; it is a refusal to write a thesis in this school. Same for "we delimit ourselves from testing", "from mathematical analysis", "from manufacturability assessment". The methods are not optional. Narrow *within* the engineering envelope: which load case, which material, which speed range, which manufacturing process, which controller variant. Not which categories of method.

If a delimitation reads like "we choose not to do engineering work on this question", the question itself is wrong and the right move is to reformulate it (see the *overpromised RQ on solid work* note in section 1.4.1), not to wave the work off in the introduction.
