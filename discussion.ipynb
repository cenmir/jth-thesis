# Discussion {#sec-discussion}

*Step back from the numbers. Discuss what the results imply for theory, for practice, and for the design choices made along the way. Be honest about what could have been done differently.*


## Theoretical implications

What does the result mean for the field? Does it confirm, extend or challenge existing theory? Be specific. "The result is interesting" is not a theoretical implication; "the result suggests that the gain-update rule converges to a stable fixed point even under sensor noise, which is not predicted by the linearised analysis in @Hu2025" is.


## Practical implications

What does the result mean for someone who would actually build one of these? Cite the requirements from the introduction and state which ones were met, which were exceeded and which were missed.


## Sustainability and ethics

*Required by the ILOs that govern societal impact and ethics (ILO 7, 10, 11). A short, substantive subsection is enough. Generic platitudes are worse than nothing.*

Cover the parts that apply to the work:

- **Material and energy footprint.** What did the manufacturing, computation or testing cost in resources? Where are the obvious reductions (a lighter design, a coarser-but-acceptable mesh, less-toxic alternatives)?
- **End-of-life and reuse.** Can the product be disassembled, recycled or refurbished? If the deliverable is a model or workflow, can it be reused beyond this thesis?
- **Stakeholders affected.** Who benefits, who carries risk, who is excluded? For an industry partner this includes operators, end users and the surrounding community, not only the company.
- **Ethical considerations specific to the method.** Human-subject data (informed consent, anonymisation), AI tools used in the analysis (disclosure, accountability for output), proprietary data (NDAs, public release of results).

Avoid the empty-sustainability paragraph that lists the seventeen UN goals and claims the project contributes to most of them. State the actual mechanism: "the redesigned bracket reduces mass by 230 g per unit; at 50 000 units per year that is 11.5 t of aluminium per year, equivalent to roughly 100 t of CO2 saved on primary production".


## Methods limitations

*Every thesis has limitations. List them honestly. Distinct from the delimitations stated in the introduction, limitations are constraints discovered along the way.*

Typical patterns:

- **Model fidelity.** The CAD model assumed uniform wall thickness, while the manufactured spring element experiences dilution at the bend radius. If the dilution is critical, the simulation overstates the spring's stiffness.
- **Parameter uncertainty.** Friction coefficients were taken from a generic dry-vs-lubricated table rather than measured on the actual material pair. Substituting the upper bound of the tabulated range changes which component disassembles first.
- **Compute time.** Each simulation took five hours, which limited the number of geometries that could be evaluated. Only a single concept was simulated, where ideally several would be compared.


## Lessons learned

*A short subsection on what you would do differently with another month. Distinct from "future work" (which is for the next researcher), this is for you.*

Examples:

- Friction coefficient turned out to dominate the disassembly result. With hindsight, measuring it directly on a small material coupon should have been the first task.
- Concept screening was performed by a single author, which biased the selection. Even one external reviewer with a Pugh matrix would have caught the weighting of spring functionality over contact status.
