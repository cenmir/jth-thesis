## Equations and cross-references

*Number every equation that is referenced from prose. Do not number throwaway derivations. Cross-reference with `@eq-label`. Use display math (`$$ ... $$`) for standalone equations and inline math (`$ ... $`) for variables in running text.*

Example, taken from fluid dynamics. Bernoulli's equation expresses the energy of an incompressible fluid at a coordinate $s$: the fluid pressure $p_s$, the kinetic energy $\tfrac{1}{2}\rho u_s^2$ and the potential energy $\rho g z_s$ all correlate. In incompressible, frictionless flow the total energy along a streamline is constant, as in @eq-bernoulli [@Storck2010].

$$
p_s + \tfrac{1}{2}\rho u_s^2 + \rho g z_s = \text{constant}
$$ {#eq-bernoulli}

Applying @eq-bernoulli at two points on the same streamline gives @eq-bernoulli-2pt:

$$
p_1 + \tfrac{1}{2}\rho \bar{u}_1^2 + \rho g z_1
= p_2 + \tfrac{1}{2}\rho \bar{u}_2^2 + \rho g z_2
$$ {#eq-bernoulli-2pt}

A second example, from controls. The PID control law combines a proportional, integral and derivative response to the error signal $e(t) = r(t) - y(t)$ between reference $r$ and measured output $y$, following the approach by Hu [@Hu2025]:

$$
u(t) = K_p\, e(t) + K_i \int_0^t e(\tau)\, d\tau + K_d \frac{de(t)}{dt}
$$ {#eq-pid}

@eq-pid is used in @sec-results to compare the fixed-gain implementation with the adaptive variant.


### Bold math for vectors and matrices

*Use `\bm{...}` for vectors and matrices. Do not use `\mathbf{...}`; it does not handle Greek letters correctly. Do not put scalars in bold.*

Example, drawn from contact mechanics [@Stromberg2012]. The potential energy of an elastic body as a function of its nodal displacement vector $\bm{d}$ is

$$
\Pi(\bm{d}) = \tfrac{1}{2}\bm{d}^T \bm{K} \bm{d} - \bm{F}^T \bm{d},
$$ {#eq-potential}

where $\bm{K}$ is the stiffness matrix and $\bm{F}$ is the external nodal force vector. Equilibrium minimises @eq-potential subject to the kinematic constraint that no node penetrates the rigid obstacle:

$$
\begin{cases}
  \underset{\bm{d}}{\min}\; \Pi(\bm{d}) \\
  \text{s.t.}\; \bm{C}_N \bm{d} - \bm{g} \leq \bm{0}
\end{cases}
$$ {#eq-equilibrium}

Here $\bm{C}_N$ is a transformation matrix of the contact normals and $\bm{g}$ is the vector of initial gaps. The Karush-Kuhn-Tucker conditions provide the optimal solution:

$$
\begin{cases}
  \bm{K} \bm{d} - \bm{F} + \bm{C}_N^T \bm{P}_N^A = \bm{0} \\
  \bm{P}_N^A \geq \bm{0} \\
  \bm{C}_N \bm{d} - \bm{g} \leq \bm{0} \\
  \bm{P}_N^A \circ (\bm{C}_N \bm{d} - \bm{g}) = \bm{0}
\end{cases}
$$ {#eq-kkt}

### Dense matrices and `\arraystretch`

Matrices that contain fractions, integrals or nested subscripts pack their rows so tightly that the result is hard to read. LaTeX exposes a single command to fix it: `\renewcommand{\arraystretch}{factor}`. The factor multiplies the default row height. A value of `1.0` is the default, `1.5` is comfortable for matrices of derivatives, `2.0` is roomy.

Two rules. First, the command must come *before* the matrix, in the same scope, so it is in effect when the rows are typeset. Second, reset it afterwards (back to `1.0`) so the relaxed row spacing does not leak into later matrices.

**Without `\arraystretch`** (default row height; fraction bars in adjacent rows almost touch):

$$
\bm{J}(\xi,\eta,\zeta) = \begin{bmatrix}
\dfrac{\partial x_h}{\partial \xi}   & \dfrac{\partial y_h}{\partial \xi}   & \dfrac{\partial z_h}{\partial \xi}   \\
\dfrac{\partial x_h}{\partial \eta}  & \dfrac{\partial y_h}{\partial \eta}  & \dfrac{\partial z_h}{\partial \eta}  \\
\dfrac{\partial x_h}{\partial \zeta} & \dfrac{\partial y_h}{\partial \zeta} & \dfrac{\partial z_h}{\partial \zeta}
\end{bmatrix}
$$

**With `\arraystretch`** set to `1.6`, then reset:

\renewcommand{\arraystretch}{1.6}

$$
\bm{J}(\xi,\eta,\zeta) = \begin{bmatrix}
\dfrac{\partial x_h}{\partial \xi}   & \dfrac{\partial y_h}{\partial \xi}   & \dfrac{\partial z_h}{\partial \xi}   \\
\dfrac{\partial x_h}{\partial \eta}  & \dfrac{\partial y_h}{\partial \eta}  & \dfrac{\partial z_h}{\partial \eta}  \\
\dfrac{\partial x_h}{\partial \zeta} & \dfrac{\partial y_h}{\partial \zeta} & \dfrac{\partial z_h}{\partial \zeta}
\end{bmatrix}
$$ {#eq-jacobian}

\renewcommand{\arraystretch}{1.0}

The source for the with-arraystretch version is just a `\renewcommand` before the `$$ ... $$` block and another `\renewcommand` after it. The same trick works for `tabular`, `array` and the various `bmatrix` and `pmatrix` siblings.

**One-shot row gap with `\\[6pt]`.** When only a single row needs breathing space (an integral, a fraction, a block boundary), `\arraystretch` is too blunt: it loosens every row in the matrix. The finer-grained alternative is `\\[6pt]` instead of a plain `\\` at the end of that one row. The bracketed value is a vertical skip in standard TeX units: `pt`, `ex`, `em`, `mm`.

Example. A block matrix where the upper 2x2 block should be visually separated from the lower block:

$$
\bm{M} = \begin{bmatrix}
m_{11} & m_{12} \\
m_{21} & m_{22} \\[10pt]
\bm{0}  & \bm{I}
\end{bmatrix}
$$ {#eq-rowgap-block}

Source for that matrix:

```latex
$$
\bm{M} = \begin{bmatrix}
m_{11} & m_{12} \\
m_{21} & m_{22} \\[10pt]
\bm{0}  & \bm{I}
\end{bmatrix}
$$
```

The `\\[10pt]` after row 2 adds 10 pt of vertical space between row 2 and row 3, while rows 1 and 2 stay at the default spacing. Compare with the same matrix using plain `\\` everywhere: the boundary between the two blocks disappears and the matrix reads as a single 3x2 grid.

### Resizing wide content with `\resizebox`

`\arraystretch` solves the cramped-rows problem. The orthogonal problem is content that overflows the page margin horizontally: a wide matrix, a wide table, a tikz diagram. LaTeX provides `\resizebox{width}{height}{content}` to scale arbitrary content to a target size. Pass `!` for either dimension to preserve aspect ratio. Typical use:

```latex
\resizebox{\textwidth}{!}{...content...}
```

This scales the content to exactly the text width and lets the height scale proportionally.

**Quarto equivalent.** Quarto's `width=` attribute on figures (`![](path){width=70%}`) handles figure scaling natively. There is no Quarto-native equivalent for scaling a *table* or an *equation*; for those, drop into raw LaTeX. Inside a Quarto markdown file or notebook, raw LaTeX goes in a fenced block tagged `{=latex}`:

````markdown
```{=latex}
\resizebox{\textwidth}{!}{$
\bm{J}(\xi,\eta,\zeta) = \begin{bmatrix}
\dfrac{\partial x_h}{\partial \xi}   & \dfrac{\partial y_h}{\partial \xi}   & \dfrac{\partial z_h}{\partial \xi}   \\
\dfrac{\partial x_h}{\partial \eta}  & \dfrac{\partial y_h}{\partial \eta}  & \dfrac{\partial z_h}{\partial \eta}  \\
\dfrac{\partial x_h}{\partial \zeta} & \dfrac{\partial y_h}{\partial \zeta} & \dfrac{\partial z_h}{\partial \zeta}
\end{bmatrix}$}
```
````

Use sparingly. Shrinking text below ~8 pt is unreadable; an over-shrunk equation in body text looks out of place. If the content is fundamentally too wide for the page, consider breaking it into two display equations, splitting the table, or rotating the page (`\begin{landscape}...\end{landscape}` from the `pdflscape` package).

### Raster versus vector graphics

When you embed a figure, you choose between two fundamentally different image representations. The choice has more impact on the final PDF than most students realise.

A **raster** image is a grid of pixels. Each pixel has a fixed colour. Zoom past the native resolution and you see the grid: stairsteps on diagonal lines, blocky text labels, blurred curves. Photographs and screenshots are raster. Common file extensions: `.png`, `.jpg`, `.bmp`, `.tiff`.

A **vector** image is a list of geometric instructions: "draw a curve from this point to that point, fill this polygon with this colour, place this text string at this position". The renderer turns those instructions into pixels at whatever resolution the output device asks for, every time the document is opened. There is no native resolution; the image stays crisp at every zoom level. Common file extensions: `.pdf`, `.svg`, `.eps`.

For a thesis, vector is almost always the right choice. The PDF is read on screens at variable zoom levels, on phones, and printed at 300 to 600 dpi. A pixelated axis label in figure 4.3 distracts every reader, every time. A vector label stays crisp forever, and the source can be edited months later (relabel, recolour, change font) without re-exporting the whole image.

The main case where raster is unavoidable is actual photographs: test rigs, prototypes, fracture surfaces, microscope images. Capture those at high resolution (1500 to 3000 px on the long edge) and embed them as `.png` (lossless) or `.jpg` (lossy, fine for photos). For everything else (CAD-derived diagrams, block diagrams, flow charts, plots, schematics, free-body diagrams) prefer vector.

**Tools that produce SVG or PDF:**

- **Inkscape** (free, open source). A dedicated vector editor. Save As → PDF or Plain SVG. Both formats embed directly in Quarto.
- **PowerPoint.** Build the diagram with native shapes, arrows and text boxes. Select the diagram, right-click, *Save as Picture...*, choose **SVG**. Works for anything assembled from native PowerPoint primitives. Do *not* paste a raster image into PowerPoint and then export: the export will keep the embedded pixels.
- **draw.io / diagrams.net** (free, web-based). Good for block diagrams and flow charts. File → Export As → PDF or SVG.
- **Matplotlib** (Python). Call `plt.savefig("name.pdf")` in the cell; Quarto embeds the vector PDF directly. This is what the executable-code section in @sec-python-quarto does.
- **TikZ** (LaTeX). Code-based vector graphics that compile alongside the rest of the document. Steeper learning curve, but the result integrates with the body's fonts and colours by default and stays in version control as plain text.
- **CAD systems.** Most CAD tools can export a drawing view to PDF or DXF. Open the PDF in Inkscape to trim and relabel before embedding.

Most engineering diagrams in a thesis can be drawn in one of these tools in under thirty minutes; the result will look professional and stay editable.

**File-size note.** Vector files for a typical engineering diagram are tens to a few hundred kilobytes, smaller than the raster equivalent at any reasonable print resolution. The exception is a vector file with thousands of small primitives (a dense plot exported as SVG, or a CAD drawing with every fillet rendered) which can balloon. In that case, export from the source tool with a coarser line tolerance, or rasterise the figure deliberately at 300 dpi.


### A short detour into typesetting

To understand why figures sometimes appear "in the wrong place", it helps to know what is producing the PDF.

**Typesetting** is the craft of arranging characters, words, lines and floating elements (figures, tables) on a page so the result is legible and pleasant to read. Done well, the reader notices nothing: the eye flows from line to line, the figure appears just where it is needed, the heading has the right amount of space around it. Done poorly, the reader notices everything: orphan lines at the top of a page, a figure half a page after it was mentioned, ragged spacing, a heading hanging at the bottom of a column.

For most of the twentieth century this craft lived with skilled human compositors. In the late 1970s, Donald Knuth, frustrated by the poor typesetting of mathematics in the second edition of *The Art of Computer Programming*, set out to solve the problem in software. The result was **TeX** (1978), a typesetting engine designed around line-breaking and math layout, with algorithms that are still the state of the art four decades later. Knuth froze TeX in 1989; its version number now asymptotically approaches $\pi$.

Writing TeX directly is slow because it operates at the level of boxes and glue, not chapters and sections. In 1984 Leslie Lamport built **LaTeX** as a macro layer on top of TeX, providing high-level commands such as `\section{}`, `\figure`, `\cite{}` so authors could describe a document by its structure and let TeX handle the typography. Modern LaTeX is maintained by the LaTeX Project Team and ships in distributions such as TeX Live, MiKTeX and TinyTeX. The actual engine used to compile a document is usually one of pdfTeX, XeTeX or LuaTeX. This template targets LuaTeX, which has good Unicode support and is the default that Quarto picks.

**Pandoc** (John MacFarlane, 2006) is the universal document converter that reads Markdown and writes LaTeX (and HTML, Word, EPUB, and many other formats). It is the layer that lets you write `![caption](path){#fig-X}` and have it become a properly captioned, cross-referenced `\begin{figure}...\end{figure}` block in LaTeX.

**Quarto** (Posit, 2022) sits on top of Pandoc. It coordinates the moving parts of a technical document: executable code cells (via Jupyter or Knitr), figure cross-references, multi-format output (PDF, HTML, EPUB, slides, websites), and project-level configuration like the `_quarto.yml` in this repository. Quarto is the successor to R Markdown and the closest thing the literate-programming world has to a default. It is what turns this notebook into the PDF you are reading.

When you render this thesis, the pipeline is: notebook source, Pandoc Markdown, LaTeX, LuaTeX, PDF.

### How LaTeX places figures

In source order, figures and tables are not placed where they are written. They are **floats**: LaTeX collects them, then chooses where to put each one according to a built-in placement algorithm. The algorithm tries to satisfy the constraints (no orphan lines, no half-empty pages, figure near its reference) while keeping the prose unbroken. This is why a figure mentioned on page 12 may end up on the top of page 13: LaTeX decided that breaking the paragraph on page 12 to fit the figure would have produced worse-looking output.

The author can give hints. A figure float accepts a position specifier built from these letters:

- `h` is *here*, where the figure appears in the source, if there is room.
- `t` is the *top* of a page.
- `b` is the *bottom* of a page.
- `p` is its own *page* of floats.

The default is `[tbp]` (top, bottom, or a float page); `[htbp]` adds *here* as a possibility. The order of the letters does not change priority; LaTeX always tries `h`, then `t`, then `b`, then `p`.

The `float` package (already loaded in `preamble.tex`) adds one more specifier:

- `H` is *HERE*, with no fallback. The figure goes exactly where you wrote it, even if that means a poorly broken page.

`H` is a hammer. It overrides the typesetting algorithm entirely. Used everywhere, it produces theses where a half-page figure sits at the bottom of an otherwise full page, leaving a wasteland of whitespace, while the reader is six pages past the place where the figure was mentioned.

### How to influence placement in this template

By default, this template sets `fig-pos: "H"` in `_quarto.yml`, so every figure is forced into its source location. This matches what most students intuitively expect ("the figure should go where I put it") but produces the spacing problems just described.

Three levers, from least to most invasive:

1. **Trust the typesetter.** Change the project default to `fig-pos: "htbp"` in `_quarto.yml`. LaTeX will then place each figure at the most readable nearby location and the document as a whole will breathe better. This is the recommended setting for the final draft.
2. **Override per-figure.** Append `fig-pos="..."` to a specific figure's attribute block to override the project default. For example, `![...](path){#fig-X fig-pos="t"}` asks for the top of a page; `fig-pos="H"` forces it here.
3. **Adjust the figure size.** A figure that does not fit will float far away. Reducing `width=60%` to `width=50%` often lets a stubborn float settle on the same page as its reference.

The rule of thumb: write the prose, let LaTeX place the figures, and only intervene when a specific figure has clearly misbehaved. Fighting the typesetter at every figure produces worse output, not better.


## Citations

*Use citation keys from `references.bib`. Quarto handles the rendering. The CSL file referenced from `_quarto.yml` controls the format. This template ships with two: `ieee.csl` (numeric, the default) and `apa-7th.csl` (author-date).*

### Markdown syntax

The same Quarto syntax is used regardless of which CSL is active. Only the rendered output changes.

- Parenthetical: `[@Braunl2022]`
- Author-as-subject (recommended when the author is the subject of the sentence): `@Braunl2022`
- Multiple sources at once: `[@Braunl2022; @Mathew2021; @Arefin2023]`
- With locator: `[@Callister2011, pp.\ 152-155]`

### What each style produces

Under **IEEE** (the default), citations are numeric. The first-cited source becomes `[1]`, the next `[2]`, and so on. Multiple-source citations collapse into a single bracketed list. For example, the sentence "we follow the approach used in earlier balancing-robot work" with `[@Braunl2022; @Mathew2021; @Arefin2023; @Olmez2022; @Hu2025]` renders as:

> we follow the approach used in earlier balancing-robot work [1], [2], [3], [4], [5]

When integrating a citation into the sentence, prefer the author-as-subject form: write "following the approach by Hu [@Hu2025]" rather than "[@Hu2025] proposes the approach". In IEEE this renders as "following the approach by Hu [5]"; in APA as "following the approach by Hu (2025)".

Under **APA 7**, the same `[@Braunl2022]` renders as (Braunl, 2022), and the multi-source example as (Braunl, 2022; Mathew et al., 2021; Arefin, 2023; Olmez et al., 2022; Hu, 2025).

### Switching styles

Edit one line in `_quarto.yml`:

```yaml
csl: ieee.csl       # default (numeric, IEEE Reference Guide)
# csl: apa-7th.csl  # author-date, APA 7th edition
```

Other CSL files from <https://www.zotero.org/styles> drop in the same way. Save the file next to `_quarto.yml` and point `csl:` at it.

### Which style to use

- **IEEE** suits math- and engineering-heavy theses. Numeric citations interleave cleanly with equations and tables, and the reference list is compact.
- **APA 7** suits theses with a strong empirical or qualitative component, where the author and year carry meaning inside the prose.

Ultimately the authors' choice, subject to the programme's submission guidelines.

### Citation, footnote or parenthetical?

Not everything that has a source belongs in the bibliography. A useful three-way split:

- **Citation** (`[@key]`): peer-reviewed papers, textbooks, technical standards. The bibliography is for sources whose academic weight you are leaning on.
- **Footnote** (`^[...]`): software libraries, GitHub repositories, project websites, protocol definitions, side remarks. These pollute the reference list and inflate the source count without adding academic weight.
- **Inline parenthetical**: common knowledge that needs no pointer (Ohm's law, atmospheric pressure at sea level).

Applying this rule shrinks the reference list and leaves real academic weight in what remains.

Cite the original source, not a tutoring website that summarises it. If only a marketing page or a Wikipedia entry is available, prefer a footnote and find a peer-reviewed source for the underlying claim.


## Notational style

*Pick one notation per quantity and apply it everywhere: body, equations, captions, tables, code. Inconsistent notation is the single most common reason an examiner has to re-read a paragraph.*

### Variables and constants

- Use math notation for every quantity that appears in equations. Write $P_\text{stab}$, not `stab_p`. Code identifiers (`stab_p`, `kp_max`) belong in listings, not in prose.
- Italic for scalars ($x$, $\sigma$, $\theta$). Bold roman for vectors and matrices ($\bm{d}$, $\bm{K}$), as covered in @sec-theory. Upright (non-italic) for operators, function names and units ($\sin$, $\log$, $\mathrm{Re}$).
- Subscripts are descriptive when they are not loop indices: $\sigma_y$ (yield), $T_\text{out}$ (outlet), $K_p$ (proportional gain). For loop indices, use $i$, $j$, $k$.

### Multiplication

- Use juxtaposition between variables: $F = ma$, not $F = m \cdot a$. Quarto's CLAUDE-style review will flag every `\cdot` between variables.
- Use $\cdot$ only between numerals ($5 \cdot 3 = 15$) or in vector dot products ($\bm{a} \cdot \bm{b}$).
- Use $\times$ only for cross products ($\bm{a} \times \bm{b}$) or to denote dimensions in non-math text ($50 \times 30$ mm).

### Units

- Always with a thin space between number and unit: `$25\,\text{N}$` renders as $25\,\text{N}$, not $25N$.
- SI units (m, kg, s, N, Pa, A, ...). Convert imperial to SI before reporting.
- Units are upright, not italic: $\text{m}$, not $m$ (which is mass). The `\text{...}` wrapper in math mode does this.
- Per-unit: pick one form and use it consistently. Either $\text{m/s}^2$ throughout or $\text{m}\,\text{s}^{-1}$ throughout, not both.

### First-use definitions

Every symbol must be defined at first use *and* listed in the nomenclature. "The stress $\sigma$ at the inner radius..." is a first-use definition; "The stress at the inner radius..." (no symbol attached) is not. The nomenclature list is the reader's quick reference; the first-use definition is what carries the meaning into the surrounding sentence.

### Common pitfalls

- **Symbol re-use across chapters.** $\alpha$ is the gravitational instability coefficient in Chapter 2 and a thermal expansion coefficient in Chapter 5. Pick one symbol per concept and rename the other.
- **Switching between equivalent forms.** $u(t)$ in one equation, $u_t$ in another for the same signal. Pick one.
- **Switching between derivative notations.** $\dfrac{de(t)}{dt}$ in @sec-theory, $\dot{e}$ in the methodology. Pick one.
- **Inconsistent bold.** $\bm{d}$ in one chapter, $\mathbf{d}$ in another. The template uses `\bm` everywhere (see @sec-theory).

The fix is mechanical. Keep a private notation table next to the references file and grep for inconsistencies before submission. A consistent thesis is faster to read; a faster-to-read thesis gets a better grade.
