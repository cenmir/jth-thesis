# Results {#sec-results}

*Report the empirical data, no more and no less. Reserve interpretation for the analysis and discussion chapters. A reader who disagreed with the interpretation should still be able to use the numbers from this chapter.*

Use tables for sets of numbers that are read across rows or columns, and figures for trends. Caption every table; reference every table and figure from the prose.

## Presenting CAD figures

CAD-derived figures often look fine in the CAD viewer and crowded in the thesis. Three rules apply to every CAD image in this chapter.

- **Background plain white.** Set the CAD render background to pure white. The CAD viewer's gradient or workshop scene looks slick in a presentation; in a black-on-white technical figure it competes with the geometry and reads as visual noise. White also prints reliably and reproduces consistently across screens.
- **Same fixed view across comparisons.** When you show before-and-after of a component, or several variants in a screening exercise, lock the camera position, the perspective and the orientation. Different angles between figures force the reader to mentally re-register the geometry, which costs attention they should spend on the engineering content.
- **Raster versus vector.** Most CAD tools export to raster (PNG) by default. Raster is acceptable IF the export covers the full screen at native resolution and is then scaled down in Quarto to roughly 70% of text width, so each printed pixel maps to several pixels of the original capture. Otherwise prefer a vector export: most CAD tools can output drawing views as PDF or SVG, or a 2D shaded view as a clean vector trace. Vector keeps edges and labels crisp at every zoom level. See @sec-theory on raster versus vector for the underlying reasoning.


## Presenting FEA results

A finite-element result figure is a thesis's most defensible engineering artefact and the easiest one to mis-present. Five rules.

- **Always include the colour bar.** Without it the figure says only "the part is colourful", not "the part is loaded". The colour bar must show the variable, its unit, and the numerical range. A stress plot without a numerical range is unreadable.
- **Fix the colour-bar range when comparing.** When two figures show the same response under different load cases, geometries or material models, the colour-bar range must be identical between them. Otherwise a "red spot" in one figure carries a different absolute value than a "red spot" in the other, and the reader compares colour intensities rather than numbers.
- **Same orientation between comparisons.** Same camera, same view direction, same coordinate-axes overlay. Mirroring or rotating between figures forces the reader to mentally re-map the geometry and they will mis-do it.
- **Use MPa (= N/mm²) for stress.** In mechanical engineering, MPa is the natural unit: yield, ultimate and allowable stresses are listed in MPa in materials data sheets and in design codes. The pure-SI form (Pa = N/m²) leaves the reader counting digits: 250 MPa is mild steel; 250 000 000 Pa is a number-soup that is easy to mis-read by a factor of a thousand. Note the equivalence in the caption: "stress in MPa (N/mm²)".
- **Dig deeper than the surface contour.** A surface contour is the start, not the result. Add an iso-surface plot for the value at risk (the locus where, say, von Mises stress exceeds 80 % of yield). Cut or slice the model to expose what happens inside, where stress concentrations actually peak; a bracket figure that shows only the outer surface tells you nothing about the internal stress around the bolt holes or under the fillet.


## Dynamic analysis: energy plots

For transient or dynamic simulations (explicit dynamics, drop tests, impact, contact-rich problems) include the energy time-history plot alongside the kinematic and stress results: kinetic energy, internal energy (elastic plus plastic), hourglass energy, contact energy, and total energy.

The energy plot is the verification half of a dynamic simulation. Total energy should be conserved (or decrease monotonically under explicit damping). Hourglass energy should stay below roughly 5 % of internal energy. A simulation that produces a clean stress result but a 30 % hourglass-energy ratio is not a valid simulation; the result is an artefact of the mesh.


## Verification versus validation

These are different things, and they fail in different ways.

- **Verification** asks: *are we solving the equations right?* Mesh convergence, time-step convergence, energy conservation, comparison to analytical limit cases, code-to-code cross-check. Verification is internal to the simulation. The cost is solver time and an extra appendix table.
- **Validation** asks: *are we solving the right equations?* A physical test of the actual part or system, compared against the simulation prediction. Validation is external: it requires hardware, instrumentation, and matched test data. The cost is real money and real time, but the payoff is a model whose predictions you can trust outside the cases you tested.

For a typical FEA thesis, verification is non-negotiable (mesh-convergence study at minimum, energy plot for dynamics). Validation is desirable when scope and budget allow. Without validation, the discussion chapter must explicitly mark the model as unvalidated and bound its claims to that scope.

A useful framework for *designing* the validation argument is the *Validation Square* [@Pedersen2000ValidationSquare], which separates the evidence into four quadrants: theoretical structural validity (the method's internal logic and assumptions), empirical structural validity (the example problems the method is shown to solve), empirical performance validity (the measured impact on real designs), and theoretical performance validity (the method's usefulness beyond the validation cases). Cite it explicitly when the thesis develops a new design method or a new simulation workflow whose value is bigger than a single case study.
