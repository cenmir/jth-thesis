# Results {#sec-results}

*Report the empirical data, no more and no less. Reserve interpretation for the analysis and discussion chapters. A reader who disagreed with the interpretation should still be able to use the numbers from this chapter.*

Use tables for sets of numbers that are read across rows or columns, and figures for trends. Caption every table; reference every table and figure from the prose.

Example: the performance of the two controllers is summarised in @tbl-pid-comparison.

| Metric                  | Fixed-gain PID | Adaptive PID | Improvement |
|:------------------------|---------------:|-------------:|------------:|
| Rise time [s]           |          0.42  |        0.31  |      26.2 % |
| Settling time [s]       |          1.85  |        1.12  |      39.5 % |
| Overshoot [%]           |         18.3   |        8.7   |      52.5 % |
| Steady-state error [deg]|          0.15  |        0.04  |      73.3 % |
| Integral squared error  |         12.40  |        5.60  |      54.8 % |

: Performance comparison of fixed-gain and adaptive PID controllers, averaged across ten runs on a level surface. {#tbl-pid-comparison}

As shown in @tbl-pid-comparison, the adaptive PID controller outperforms the fixed-gain controller on every metric. The PID structure in @fig-pid-controller and the control law @eq-pid form the shared basis of both controllers; only the gain-update rule differs.

Report enough decimal places to distinguish meaningful differences, but no more. Two to three significant figures is usually right.
