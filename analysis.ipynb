## RQ1

*Restate RQ1 here, then answer it using the evidence in @sec-results.*

> **RQ1:** How do the rise time, settling time, overshoot and steady-state error of the bio-inspired adaptive PID controller compare with those of a Ziegler-Nichols-tuned fixed-gain controller, across an operating envelope spanning 0 to 1 m/s forward velocity?

Template paragraph: across the ten matched runs summarised in @tbl-pid-comparison, the adaptive controller reduced rise time from 0.42 s to 0.31 s (26 %), settling time from 1.85 s to 1.12 s (40 %), and overshoot from 18.3 % to 8.7 % (52 %). Steady-state error fell from 0.15° to 0.04° (a factor of 3.75). The improvements held across the velocity envelope: at 1 m/s the settling time of the adaptive controller was 1.18 s, compared with 2.04 s for the fixed-gain controller. Run-to-run variability also tightened, with the standard deviation of settling time at 0.04 s for the adaptive variant against 0.09 s for the fixed-gain variant.


## RQ2

*Restate RQ2 here, then answer it using the evidence in @sec-results.*

> **RQ2:** By how much does the adaptive controller reduce the peak tilt angle and the recovery time after a 15° impulse disturbance, compared with the fixed-gain controller, when both run at 200 Hz on the target MCU?

Template paragraph: in the impulse-disturbance test described in @sec-methodology, the adaptive controller bounded the peak tilt angle at 6.2°, against 9.8° for the fixed-gain controller (a 37 % reduction). Recovery to within $|\theta| < 5°$ took 0.84 s on average for the adaptive controller, versus 1.71 s for the fixed-gain controller (a 51 % reduction). Three out of ten fixed-gain runs exceeded the recovery threshold at the end of the 2 s observation window; all ten adaptive runs returned to threshold within 1.1 s.
